```python
s = input('Enter the string')
term = input('Which term would you like to search')

counter = 0

for i in s:
    if i == term:
        counter +=1

print('Frequency',counter)
```

```python
s = input('Enter the number of string')

counter = 0
for i in s:
    counter += 1

print(counter)
```

```python
# To calculate the length of the string 
# without using the len function
name = input('Enter the number of strings to count:')
counter = 0
for i in name:
    counter += 1

print('The length of the string is:', counter)
```

```python
s = input('Enter the string')
term = input('What would you like to search for')

counter = 0
for i in s:
    if i == term:
        counter += 1

print('Frequency', counter)
```

```python
# Write a program which can remove a particular character from a string
s = input('Enter the string')
term = input('What would you like to remove')

result = ''
for i in s:
    if i != term:
        result = result + i

print(result)
```
